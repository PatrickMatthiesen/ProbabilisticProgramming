## Exercise 11E1

Compute the answer to the exercise in the cell below

```python
# If an event has probability 0.35, what are the log-odds of this event?

propability = 0.35
log_odds = np.log(propability / (1 - propability))
print(f"Log-odds of event with probability {propability} is {log_odds:.2f}")
```

## Exercise 11E2

Compute the answer to the exercise in the cell below

```python
# If an event has log-odds 3.2, what is the probability of this event?

log_odds = 3.2
probability = np.exp(log_odds) / (1 + np.exp(log_odds))
print(f"Probability of event with log-odds {log_odds} is {probability:.2f}")
```

Why do Poisson regressions sometimes require the use of an offset? Provide an example

If we are using Poisson regression to model rates instead of counts, then it will be necessary to use an offset. This is because the Poisson distribution is defined for counts, and if we are modeling rates, we need to adjust for the exposure time. The offset allows us to account for the different exposure times in our model.

For example, if we are modeling the number of accidents per month in different cities, we may have different numbers of days in each month. In this case, we can use the logarithm of the number of days as an offset in our Poisson regression model. This will allow us to model the rate of accidents per day instead of the total number of accidents.

Mathematicaly we can describe the difference between 3 days and 30 days where the meaningfullness of the offset is clear. The offset is the log of the number of days, and we can see that the model will be different for 3 days and 30 days. The offset allows us to account for the different exposure times in our model.
This means shorter time periods will have a higher rate of accidents, and longer time periods will have a lower rate of accidents. The offset allows us to account for this difference in exposure time, and it is important to include it in our model to get accurate results.

If a coefficient in a Poisson regression has value 1.7, what does this imply about the change
in the outcome?

Compared to logistic regression, the coefficient in a Poisson regression is interpreted as the change in the log of the expected value of the outcome. 
So where logistic regression is interpreted as the change in the log odds of the outcome, Poisson regression is interpreted as the change in the log of the expected value of the outcome.
This means that the coefficient of 1.7 in a Poisson regression model indicates that for a one unit increase in the predictor variable, the **expected value** of the outcome will increase by a factor of exp(1.7) = 5.44. This means that the expected value of the outcome will be 5.44 times larger for a one unit increase in the predictor variable.

Explain why the logit link is appropriate for a binomial generalized linear model.

Simply it ensures that the value is beween 0 and 1. 

The logit link is appropriate for a binomial generalized linear model because it allows us to model the probability of success as a function of the predictor variables. The logit link function transforms the probability of success (p) into the log odds (log(p/(1-p))), which is a linear function of the predictor variables. This transformation allows us to model the relationship between the predictor variables and the probability of success in a way that is more interpretable and easier to work with than modeling the probability directly. Additionally, the logit link function ensures that the predicted probabilities are always between 0 and 1, which is important for a binomial outcome.
